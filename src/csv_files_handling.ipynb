```python
# ── Import the weekly summaries
#  ────────────────────────────────────────────────────────────

base = Path(r"C:\Users\chris\OneDrive - Danmarks Tekniske Universitet\Documents\DTU\Thesis\Green Power and KONSTANT\github repo")

input_path_rf = base / "DST data" / "rf_prediction" / "rf_weekly_summary.csv"
input_path_xgb = base / "DST data" / "xgb_prediction" / "xgb_weekly_summary.csv"

csv_rf = pl.read_csv(input_path_rf)
csv_xgb = pl.read_csv(input_path_xgb)

csv_rf.write_parquet(input_path_rf.with_suffix(".parquet"))
csv_xgb.write_parquet(input_path_xgb.with_suffix(".parquet"))

print("CSV files have been converted to Parquet format")
```

```python
# ── Import and merge the weekly predictions
base = Path(r"C:\Users\chris\OneDrive - Danmarks Tekniske Universitet\Documents\DTU\Thesis\Green Power and KONSTANT\github repo")

rf_csv_folder = base / "DST data" / "rf_prediction"
xgb_csv_folder = base / "DST data" / "xgb_prediction"

rf_output_path = base / "DST data" / "rf_prediction" /"rf_merged_predictions.parquet"
xgb_output_path = base / "DST data" / "xgb_prediction" /"xgb_merged_predictions.parquet"

lf_rf = pl.scan_csv(
    f"{rf_csv_folder}/file_*.csv",
    infer_schema_length=10000
)

lf_xgb = pl.scan_csv(
    f"{xgb_csv_folder}/file_*.csv",
    infer_schema_length=10000
)

lf_rf.sink_parquet(rf_output_path)
lf_xgb.sink_parquet(xgb_output_path)
```
